```python
summary_template = """
    Give the information {person_information} about a person. I want to create a short summary about the person
"""
```

```python
summary_prompt_template = PromptTemplate(
    input_variables=["person_information"],
    template=summary_template
)

# temparture is set as 0. we need non creative answer
llm = ChatOpenAI(temperature=0, model = model_name)
```

```python
# we are using langChain expression language(LCEL) to build the chain(Runnable Chain)
chain = summary_prompt_template | llm

# response is a AIMessage class's object
response = chain.invoke(input={"person_information":person_information_details})
```
